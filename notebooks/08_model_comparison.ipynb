```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

tuned_results = pd.read_csv(
    "../reports/tuned_model_comparison.csv"
)
```

```python
tuned_results.round(4)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=tuned_results,
    x="Model",
    y="F1"
)

plt.title("F1 Score Comparison")
plt.xlabel("Model")
plt.ylabel("F1 Score")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig(
    "../reports/figures/model_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1", "ROC-AUC"]

comparison = tuned_results.set_index("Model")[metrics]

comparison.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Model Performance Comparison")
plt.xlabel("Model")
plt.ylabel("Score")
plt.xticks(rotation=20)
plt.legend(title="Metric")
plt.ylim(0, 1)
plt.tight_layout()

plt.savefig(
    "../reports/figures/all_model_metrics.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
